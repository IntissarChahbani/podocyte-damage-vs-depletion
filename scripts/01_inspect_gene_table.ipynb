```python
import pandas as pd

df = pd.read_csv("data/reference/pds_genes.csv")
df.columns = df.columns.str.strip()          # removes the stray space in the header

df["mouse_gene_symbol"] = df["mouse_gene_symbol"].str.strip()
df["direction"] = df["direction"].str.strip().str.lower()

assert len(df) == 42, f"expected 42 genes, got {len(df)}"
assert df["mouse_gene_symbol"].is_unique, "duplicate gene symbols"
assert set(df["direction"]) == {"down", "up"}, "unexpected direction values"
print(df["direction"].value_counts())
print("OK")
```
